```python

from pathlib import Path
import sys

# Locate ForecastGuard project directory
project_root = Path.cwd().resolve()

while not (
    project_root / "data/selected_product_sales.csv"
).exists():
    parent = project_root.parent

    if parent == project_root:
        raise FileNotFoundError(
            "ForecastGuard project folder not found"
        )

    project_root = parent

# Allow notebook to import our Python module
sys.path.insert(0, str(project_root / "src"))

from data_pipeline import (
    prepare_product_data,
    FEATURES
)

# Run the reusable pipeline
df = prepare_product_data(project_root)

print("Dataset rows:", len(df))
print("Products:", df["item_id"].nunique())
print("Missing-price rows:", (~df["price_available"]).sum())
print("Features:", FEATURES)

assert len(df) == 9565
assert df["item_id"].nunique() == 5
assert (~df["price_available"]).sum() == 336

print("\nAll initial pipeline checks passed!")
```
